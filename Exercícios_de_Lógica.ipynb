1. Faça um algoritmo para ler dois números e imprimir a soma dos números lidos.

## Algoritmo Soma

1. Leia o primeiro número e guarde numa variável chamada numero1
2. Leia o segundo número e guarde em outra variável chamada numero2
3. Faça a soma das duas variáveis e guarde o resultado em outra variável chamada soma
4. Mostre o resultado

```python
numero1 = int(input('Informe o valor do 1° número: '))
numero2 = int(input('Informe o valor do 2° número: '))
soma = numero1 + numero2
print(f'O resultado da soma é {soma}')
```

## Algoritmo 8

Faça um algoritmo para ler três números e imprimir se estes podem ou não formar um triângulo.

Observação – Para formar os lados de um triângulo cada um dos valores tem que ser menor que a soma dos outros dois.

```python
numero1 = int(input('Insira o 1° número: '))
numero2 = int(input('Insira o 2° número: '))
numero3 = int(input('Insira o 3° número: '))

if numero1 >= numero2 + numero3 or numero2 >= numero1 + numero3 or numero3 >= numero1 + numero2:
  print('Não é um triângulo')
else:
  print('É um triângulo')
```

## Algoritmo 9

Faça um algoritmo para ler três números e se estes poderem formar um triângulo dizer se o triângulo é “EQUILÁTERO”, “ISÓSCELES” OU “ESCALENO”.

```python
numero1 = int(input('Insira o 1° número: '))
numero2 = int(input('Insira o 2° número: '))
numero3 = int(input('Insira o 3° número: '))

if numero1 >= numero2 + numero3 or numero2 >= numero1 + numero3 or numero3 >= numero1 + numero2:
  print('Não é um triângulo')
else:
  if numero1 == numero2 == numero3:
    print('É um triângulo equilátero')
  elif numero1 == numero2 or numero1 == numero3 or numero2 == numero3:
    print('É um triângulo isósceles')
  else:
    print('É um triângulo escaleno')
```
